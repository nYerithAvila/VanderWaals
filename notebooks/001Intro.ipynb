```python
#Importacion de librerias 

import pandas as pd 
import plotly.express as px
```

```python
# Definir ruta 
ar = pd.read_excel("../data/Agua_Renovable_EntidadFederativa_2020.xlsx")
ar.head()
```

```python
#Limpieza y cambio de nombres
ar_plot = ar.drop(columns=["Año", "Clave Entidad Federativa", "Agua Renovable (hm³/año)", "Población (Millones de habitantes)", "N.P."])
ar_plot = ar_plot.rename(columns={
    "Entidad Federativa": "entidad_federativa",
    "Agua Renovable per cápita (m³/habitante/año)": "agua_reno_percapita"
})
ar_plot.head(5)
```

```python
#Gráfica con Plotly
fig = px.bar(
    ar_plot, 
    x="entidad_federativa", 
    y="agua_reno_percapita", 
    title="Agua renovable per capita (m3/persona al año)",
    color= 'agua_reno_percapita', 
    labels= {"agua_reno_percapita" : "Agua Renovable",
             "entidad_federativa" : ""
        
    }
)
fig.show()
```
